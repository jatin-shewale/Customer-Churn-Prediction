```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
## Load the trained model, scaler pickle, onehot 

model = load_model("model.h5")

## Load the encoder and scaler

with open('label_encoder_gender.pkl', 'rb') as file:
    label_gender = pickle.load(file)

with open('onehot_encoder_geo.pkl', 'rb') as file:
    label_geo = pickle.load(file)

with open('scaler.pkl', 'rb') as file:
    scaler = pickle.load(file)
```

```python
## Example Input

input_data = {
    "CreditScore" : 600,
    "Geography" : "France",
    "Gender" : "Male",
    "Age" : 40,
    "Tenure" : 3,
    "Balance" : 60000,
    "NumOfProducts" : 2,
    "HasCrCard" : 1,
    "IsActiveMember" : 1,
    "EstimatedSalary" : 50000
}
```

```python
geo_encoded = label_geo.transform(
    [[input_data["Geography"]]]
)

# Convert encoded values into DataFrame
geo_encoded_df = pd.DataFrame(
    geo_encoded,
    columns = label_geo.get_feature_names_out(['Geography'])
)
```

```python
geo_encoded_df.head()
```

```python
input_data = pd.DataFrame([input_data])
```

```python
input_data
```

```python
## Encode gender

input_data['Gender'] = label_gender.transform(input_data['Gender'])
```

```python
input_data.head()
```

```python
## Combine one-hot encoded columns with input data

input_data = pd.concat([input_data.drop("Geography", axis = 1), geo_encoded_df], axis = 1)
```

```python
input_data
```

```python
## Scaline the input

input_scaled = scaler.transform(input_data)
```

```python
input_scaled
```

```python
## Predict churn

my_prediction = model.predict(input_scaled)
```
